```python
import numpy as np
from sklearn.linear_model import LogisticRegression
rng = np.random.default_rng(0)
X = rng.normal(size=(100, 2)); y = (X[:, 0] > 0).astype(int)
print('exactitud:', round(LogisticRegression().fit(X, y).score(X, y), 2))
```
